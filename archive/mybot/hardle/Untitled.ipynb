```python
word_guess = "allay"
word = "llaad"
use_answer = []
use_guess = []
green = 0
yellow = 0
for i in range(length):
    if(word[i] == word_guess[i]):
        use_guess.append(True)
        use_answer.append(True)
        green += 1
    else:
        use_guess.append(False)
        use_answer.append(False)
for i in range(length):
    if(use_guess[i]):
        continue
    for j in range(length):
        if(use_answer[j]):
            continue
        if(word_guess[i] == word[j]):
            use_answer[j] = True
            yellow += 1
            break
```
